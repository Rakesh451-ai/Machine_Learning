```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# ensamble --> bagging
```

```python
df = pd.read_csv(os.path.join(path,"diabetes_data.csv"))
```

```python
# pick some columns --> Features selection
columns_to_pick = [
    "Insulin_Levels",
    "OGTT",
    "Creatinine_Levels",
    "Random_Blood_Glucose",
    "LDL_Cholesterol",
    "Uric_Acid_Levels"
]
```

```python
X = df[columns_to_pick]
Y = df["Diabetes_Status"]
```

```python
X_train , X_test , Y_train , Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
# 100 -- random trees

model = RandomForestClassifier(
    n_estimators=300,
    max_depth=5,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=22,
    n_jobs = -1
    )
```

```python
# labelencoder --> to change string to numerical data type
# positive -> 0
# negative -> 1
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
Y_train_encoded = le.fit_transform(Y_train)
# fit --> train model on trained data -> to remember data to model
Y_test_encoded = le.transform(Y_test)
# not write fit -> we not want to remember test data to model

model.fit(X_train,Y_train_encoded)
```

```python
Y_pred = model.predict(X_test)
accuracy = accuracy_score(Y_test_encoded,Y_pred)
print("Accuracy:",accuracy)
```

```python
print(classification_report(Y_test_encoded,Y_pred))
```

```python

model.fit(X_train, Y_train_encoded) # (training_data , Output_data)
```

```python
# to know about which features is important
importance = pd.Series(
    model.feature_importances_,
    index=X.columns
)
importance.sort_values(ascending=False)
```
